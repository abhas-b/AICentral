# Key Distributions:

## Binomial Distribution:

* Models number of successes in n independent trials. Each trial has the same probability p of success.
* Can be used to model classification accuracy.

## Poisson Distribution:

* models number of events happening in a fixed interval of time or space.
* eg: number of clicks on an ad or system failures.
* The events are assumed to be independent and at a constant rate lambda.
* Used to model count data where events occur randomly over time or space.
* Can be used for anomaly detection by comparing observed counts to expected rates.
* Modeling rare events.

## Normal Distribution

* models naturally occurring data
* eg: test scores, sensor noise, errors in measurements, distribution of model residuals.
* Central Limit Theorem guarantees sums of independent variables tends towards normality.
* Used in regression, clustering, anomaly detection, Bayesian methods.

## Exponential Distribution

* models waiting times between events in a Poisson process.
* eg: time between user visits or system failures.
* Can model expected time between rare events.

## Uniform Distribution

* all outcomes in the range are equally likely.
* useful to start with when you dont have any prior information.

## Beta Distribution

* It is a continuous distribution defined on the interval [0,1][0,1][0,1].
* Used to model probabilities and proportions.
* Shape parameters: alpha, beta.
* Commonly used as a prior distribution in Bayesian stats for probablities.
* Models uncertainty about probabilities.


## Gamma Distribution

* It is a continuous distribution defined on [0, infinity).
* Controlled by by two parameters:
    * Shape k (or alpha)
    * Scale theta (inverse of rate lambda)

* It generalizes the Exponential distribution by modeling waiting times for multiple events.
* Models waiting times until the k-th event in a Poisson process.
Used in Bayesian modeling as a prior for rates and variances.
* Used in modeling skewed positive valued data such as insurance claims, rainfall, failure times.

# Binomial Distribution

$$P(X=k) = \binom{n}{k}p^k(1-p)^{(n-k)}$$

```python
from scipy.stats import binom
import numpy as np
import matplotlib.pyplot as plt
```

```python
n = 30 #number of trials
p = 0.4 #success probability

x = np.arange(0, n+1)
```

```python
pmf = binom.pmf(x,n,p)


plt.bar(x,pmf, color='blue', alpha=0.7)
plt.xlabel('Number of successes')
plt.ylabel('Probablity')
plt.title(f'Binomial Distribution n={n}, p={p}');
```

# Poisson Distribution

$$P(X=k) = \frac{\lambda^ke^{-\lambda}}{k!}$$

```python
from scipy.stats import poisson
```

```python
lambda_ = 5 #rate
x = np.arange(0,15)

pmf = poisson.pmf(x, lambda_)


plt.bar(x,pmf, color='blue', alpha=0.7)
plt.xlabel('Number of Events')
plt.ylabel('Probablity')
plt.title(f'Poission Distribution lambda = {lambda_}');
```

# Normal Distribution

$$f(x) = \frac{-exp(\frac{-(x-\mu)^2}{2\sigma^2})}{\sqrt{2\pi\sigma^2}}$$

```python
from scipy.stats import norm

mu, sigma = 0, 1
x = np.linspace(mu-4*sigma, mu+4*sigma, 1000)
pdf = norm.pdf(x, mu, sigma)

plt.plot(x, pdf, color='red');
plt.xlabel('x')
plt.ylabel('Density');
```

# Exponential Distribution

$$f(x) = \lambda e^{-\lambda x}$$

```python
from scipy.stats import expon

lambda_ = 1.5
x = np.linspace(0, 4, 1000)

pdf = expon.pdf(x, scale=1/lambda_)


plt.plot(x, pdf, color='blue');
```

# Continuous Uniform Distribution



```python
from scipy.stats import uniform

a,b = 0,10
x = np.linspace(a-1, b+1, 1000)
pdf = uniform.pdf(x, loc=a, scale=b-a)

plt.plot(x, pdf);
```

# Beta Distrbution

$$f(x, \alpha, \beta) = \frac{x^{\alpha - 1}(1-x)^{\beta - 1}}{B(\alpha, \beta)}$$

```python
from scipy.stats import beta

x = np.linspace(0,1,1000)

params = [
    (0.5,0.5), # U shaped
    (1,1),     # Uniform
    (2,2),     # symmetric, bell shaped
    (5,1),     # skewed right
    (1,3)      # skewed left
]

plt.figure(figsize=(10,6))

for a, b in params:
    y = beta.pdf(x, a, b)
    plt.plot(x, y, label=f'alpha={a}, beta={b}')

plt.legend();
```

# Gamma Distribution

$$f(x, k, \theta) = \frac{x^{k-1}e^{\frac{-x}{\theta}}}{\theta^k \Gamma(k)}$$

```python
from scipy.stats import gamma

x = np.linspace(0, 20, 1000)

params = [
    (1,1),
    (1,2),
    (2,2),
    (5,1),
    (9,0.5)
]

plt.figure(figsize=(10,6))

for shape, scale in params:
    plt.plot(x, gamma.pdf(x, shape, scale), label=f'Shape={shape}, Scale={scale}')

plt.legend();
```
