# Confidence Intervals

Point estimates of mean and standard deviation.


Key parameters of a CI: number of samples and variance.

Eg:

$$\hat\mu = \frac{\sum_{i=1}^{N} x_i}{N}$$

$$\hat\sigma^2 = \frac{\sum_{i=1}^{N}(x-\bar x)^2}{N-1}$$

Lower bound (assuming standard normal distribution):

For a gamma confidence interval, we need:

inverse CDF(gamma) = threshold

eg: inverse CDF(0.025) = -1.96


Upper bound:

CDF(gamma) = threshold

eg:
CDF(0.975) = 1.96

# Sums of Normals

if 

$$X_1 = N(\mu_1, \sigma_1^2)$$
$$X_2 = N(\mu_2, \sigma_2^2)$$

then 

$$X_1 +X_2 = N(\mu_1 + \mu_2, \sigma_1^2 + \sigma_2^2)$$

if X1 and X2 are independent

If Xi's are IID:

then 

$$X = N(\mu, \sigma^2)$$

$$\sum X_i = N(N\mu, N\sigma^2)$$

$$\hat \mu = N(\mu, \frac{\sigma^2}{N})$$

This is true regardless of data distribution, by virtue of Central Limit Theorem.



Therefore: sample mean and variance: 

$$\hat \mu = \frac{\sum x_i}{N} = \mu$$

$$\sigma_0^2 = var(\hat \mu) = \frac{\sigma^2}{N}$$


* 95% Confidence Interrval:

$$\mu: [\hat \mu - 1.96\frac{\hat \sigma}{\sqrt N}, \hat \mu + 1.96\frac{\hat \sigma}{\sqrt N}]$$

* A 95% confidence interval means that 95% of the time, the 95% confidence interval will contain the true parameter. 

If your distribution is not standard normal, then transform it to get z:
$$z = \frac{x - \mu}{\sigma_0}$$

Hence, for non-standard normal:

$$\hat \mu  - 1.96\sigma_0 <= \mu <= \hat \mu + 1.96\sigma_0$$

For sample mean, since sigma_0^2 = sigma^2/N,

$$CI = [\hat \mu - \frac{1.96\sigma}{\sqrt N}, \hat \mu + \frac{1.96\sigma}{\sqrt N}]$$


In practice, we use $\hat{\sigma}$ instead of $\sigma$ since the latter is not available. We only have the sample data and not the population.

More accurate way:

$$CI = [\hat \mu + t_{left}\frac{\hat \sigma}{\sqrt N}, \hat \mu + t_{right}\frac{\hat \sigma}{\sqrt N}]$$

This takes into the fact that $\sigma$ is unknown.

For large samples, this is almost same.

```python
import numpy as np
from scipy.stats import norm, t
```

```python
N = 1000
mu = 5
sigma = 2


data = norm.rvs(loc=mu, scale=sigma, size=N)
```

```python
mu_hat = np.mean(data)
sigma_hat = np.std(data, ddof=1)
z_left = norm.ppf(0.025)
z_right = norm.ppf(0.975)

ci_norm_lower = mu_hat+z_left*sigma_hat/np.sqrt(N)
ci_norm_upper = mu_hat+z_right*sigma_hat/np.sqrt(N)

print(f'Mu Estimate: {mu_hat:.2f}, 95% CI: [{ci_norm_lower:.2f}, {ci_norm_upper:.2f}]')
```

```python
mu_hat = np.mean(data)
sigma_hat = np.std(data, ddof=1)
t_left = t.ppf(0.025, df=N-1)
t_right = t.ppf(0.975, df=N-1)

ci_t_lower = mu_hat+t_left*sigma_hat/np.sqrt(N)
ci_t_upper = mu_hat+t_right*sigma_hat/np.sqrt(N)

print(f'Mu Estimate: {mu_hat:.2f}, 95% CI: [{ci_t_lower:.2f}, {ci_t_upper:.2f}]')
```
