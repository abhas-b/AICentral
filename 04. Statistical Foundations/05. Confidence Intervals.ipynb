# Confidence Intervals

Point estimates of mean and standard deviation.

Eg:

$$\hat\mu = \frac{\sum_{i=1}^{N} x_i}{N}$$

$$\hat\sigma^2 = \frac{\sum_{i=1}^{N}(x-\bar x)^2}{N-1}$$

Lower bound (assuming standard normal distribution):

For a gamma confidence interval, we need:

inverse CDF(gamma) = threshold

eg: inverse CDF(0.025) = -1.96


Upper bound:

CDF(gamma) = threshold

eg:
CDF(0.975) = 1.96

# Sums of Normals

if 

$$X_1 = N(\mu_1, \sigma_1^2)$$
$$X_2 = N(\mu_2, \sigma_2^2)$$

then 

$$X_1 +X_2 = N(\mu_1 + \mu_2, \sigma_1^2 + \sigma_2^2)$$

if X1 and X2 are independent

If Xi's are IID:

then 

$$X = N(\mu, \sigma^2)$$

$$\sum X_i = N(N\mu, N\sigma^2)$$

$$\hat \mu = \frac{\sum x_i}{N}$$

Therefore:

$$\hat \mu = N(\mu, \frac{\sigma^2}{N})$$

* 95% Confidence Interrval:

$$\mu: [\hat \mu - 1.96\frac{\hat \sigma}{\sqrt N}, \hat \mu + 1.96\frac{\hat \sigma}{\sqrt N}]$$

* A 95% confidence interval means that 95% of the time, the 95% confidence interval will contain the true parameter.
