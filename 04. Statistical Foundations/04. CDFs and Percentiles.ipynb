```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
mu = 170
sd = 7
N = 100
```

```python
#Generate samples from a normal distribution basis above parameters
x = norm.rvs(loc=mu, scale=sd, size=N)
```

```python
# MLE Estimate of mean

x.mean()
```

```python
#MLE Variance estimate
x.var()
```

```python
#MLE Standard deviation estimate
x.std()
```

```python
#Unbiased variance and sd using ddof: delta degrees of freedom
x.var(ddof=1), x.std(ddof=1)
```

```python
# which value sits at 95th percentile

norm.ppf(0.95, loc=mu, scale=sd)
```

```python
#give me 95th percentile

norm.cdf(181.5, loc=mu, scale=sd)
```
